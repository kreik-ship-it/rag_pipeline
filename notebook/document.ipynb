### Data Ingestion 


```python
from langchain_core.documents import Document
from langchain_community.document_loaders import PyPDFLoader
from langchain_community.document_loaders import DirectoryLoader
from collections import Counter 
from pathlib import Path
```

```python
loader = PyPDFLoader("../data/2021-02-17_cc_08-2021_transformation_stromerzeugung_2050_0.pdf")
documents = loader.load()
```

```python
# Load all pdf documents from this path 

dir_loader = DirectoryLoader(
    "../data",
    glob="*.pdf",
    loader_cls=PyPDFLoader,
    show_progress=True
)

documents = dir_loader.load()

```

```python
sources = Counter(doc.metadata["source"] for doc in documents)

print(f"Anzahl verschiedener PDFs: {len(sources)}")

for source, count in sources.items():
    print(f"- {Path(source).name}: {count} Seiten")
```
